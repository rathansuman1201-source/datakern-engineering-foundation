## 1. Exploring Orders (`olist_orders_dataset.csv`)

```python

import pandas as pd
import os

#Load raw orders dataset
DATA_FILE = os.path.join(os.getcwd(), "..", "data", "raw","olist_orders_dataset.csv")
df = pd.read_csv(DATA_FILE)

print("Display by default:")
print(df.head())
print("Display first 10 rows:")
print(df.head(10))
print("Display last by default:")
print(df.tail())

print("Dataset size:\n", df.shape)
print("\nPrint info:\n")
df.info()

print("Summary Statistics:\n")
print(df.describe(include="all"))
print("\nMissing values:\n")
print(df.isna().sum())

print("\nDuplicate rows:\n")
print(df.duplicated().sum())


print("\nTypes of Order status\n", df["order_status"].unique())
print("\nNumber of unique order status:\n", df["order_status"].nunique())
print("\nOrder status counts:\n", df["order_status"].value_counts())

```
