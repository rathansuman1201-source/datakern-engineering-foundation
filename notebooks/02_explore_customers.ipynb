# 2. Exploring Customers (`olist_customers_dataset.csv`)

```python
import pandas as pd
import os

#Load raw orders dataset
DATA_FILE = os.path.join(os.getcwd(), "..","data","raw","olist_customers_dataset.csv")
df = pd.read_csv(DATA_FILE)

#Inspect a few records — head() and tail() 
print("Display first by default:\n", df.head())
print("\nDisplay first 10 only:\n",df.head(10))
print("\nDisplay last by default:\n",df.tail())

#Understand the size — shape
print("\nDataset size\n", df.shape)

#Inspect columns and data types — info()
print("\nDisplay info:\n",df.info())


#Review summary statistics — describe()
print("\nSummary Statistics:\n", df.describe(include="all"))

#Check missing values — isna().sum()
print("\nPrint missing values:\n", df.isna().sum())

#Check potential duplicate rows — duplicated()
print("\nPrint duplicate values:\n",df.duplicated().sum())

#Explore distinct values and categories
print("\nList of cities\n", df["customer_city"].unique())
print("\nNumber of cities\n", df["customer_city"].nunique())
print("\nList of Cities:\n", df["customer_city"].value_counts())
print("\nNumber of unique customer ids\n", df["customer_id"].nunique())
print("\nNumber of unique customer zip code prefix\n", df["customer_zip_code_prefix"].nunique())


#Select and inspect specific rows — .loc[] and .iloc[]
print("\nDisplay all the details of a customer through index\n", df.loc[0])
print("\nDisplay only the city details of a customer through index\n", df.loc[0,'customer_city'])

print("\nDisplay all the details of a customer via integer position\n", df.iloc[0])
print("\nDisplay only the city details of a customer via integer position\n", df.iloc[0,3])


```
